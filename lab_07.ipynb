```python
import re 
import nltk
import pymorphy3

from nltk.corpus import stopwords
nltk.download("stopwords")

morph = pymorphy3.MorphAnalyzer()

stop_words = set(stopwords.words("russian"))

stop_words.update([
    "это", "такой", "также", "который", "свой",
    "можно", "ли", "какой", "какие", "какая",
    "какое", "сколько", 'ниу', 'вшэ'
])

def preprocess_text(text): 
    text = text.lower()
    tokens = re.findall(r"[а-яё]+", text)

    lemmas = [
        morph.parse(word)[0].normal_form
        for word in tokens
    ]

    lemmas = [
        word for word in lemmas
        if word not in stop_words and len(word) > 1
    ]

    return lemmas
```

## Тестовые вопросы

Для оценки качества системы готовим два набора: вопросы, ответы на которые
есть в базе знаний, и вопросы по той же теме, ответов на которые в базе нет.

```python
test_questions = [
    # Вопросы, ответы на которые есть в базе знаний (10)

    "Сколько республик было в Советском Союзе на момент распада?",
    "В каком году образовался Советский Союз?",
    "Кто возглавлял КПСС накануне распада Союза?",
    "Когда Германия напала на Советский Союз?",
    "Как называется курс реформ, начатый Горбачёвым в 1985 году?",
    "Когда советские войска ушли из Афганистана?",
    "Кто первым полетел в космос?",
    "Почему в советских магазинах не хватало товаров?",
    "Что подписали в Беловежской пуще в декабре 1991 года?",
    "Как расшифровывается название СССР?",

    # Вопросы, ответов на которые в базе знаний нет (5)

    "Какой гимн был у СССР?",
    "Сколько стоила буханка хлеба в СССР?",
    "Какие курорты были популярны у советских туристов?",
    "Как была устроена школьная система образования в СССР?",
    "Какие фильмы снимали на киностудии «Мосфильм»?"
]

# Индексы правильных ответов для первых десяти вопросов.
# Для последних пяти правильного ответа в базе нет.
correct_indices = [
    3, 1, 8, 12, 20, 23, 17, 22, 25, 0
]
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def jaccard_similarity(tokens1, tokens2): 
    set1 = set(tokens1)
    set2 = set(tokens2)

    union = set1 | set2

    if len(union) == 0: 
        return 0.0

    intersection = set1 & set2

    return len(intersection) / len(union)


def cosine_similarities(query_vector, database_vectors): 
    return cosine_similarity(
        query_vector.reshape(1, -1),
        database_vectors
    )[0]

def pearson_similarity(query_vector, database_vectors): 
    scores = []

    for vector in database_vectors: 
        if np.std(query_vector) == 0 or np.std(vector) == 0: 
            scores.append(0.0)
        else: 
            correlation = np.corrcoef(query_vector, vector)[0,1]

            if np.isnan(correlation): 
                correlation = 0.0

            scores.append(correlation)

    return np.array(scores)
```

```python
NO_ANSWER = "Ответ не найден"

def find_answers(query, top_n=3):
    processed_query = preprocess_text(query)

    query_vector = vectorizer.transform([processed_query]).toarray()[0]

    jaccard_scores = np.array([
        jaccard_similarity(processed_query, question)
        for question in processed_questions
    ])

    cosine_scores = cosine_similarities(query_vector, X)

    pearson_scores = pearson_similarity(query_vector, X)

    scores_by_metric = {
        "Жаккар": jaccard_scores, 
        "Косинусное сходство": cosine_scores, 
        "Корреляция Пирсона": pearson_scores
    }

    results = {}

    for metric_name, scores in scores_by_metric.items(): 
        results[metric_name] = []

        # Ни одного общего слова с базой: честно сообщаем, что ответа нет,
        # а не возвращаем случайный вопрос с нулевым сходством
        if scores.max() <= 0:
            results[metric_name].append({
                "index": None,
                "score": 0.0,
                "matched_question": None,
                "answer": NO_ANSWER
            })
            continue

        top_indices = np.argsort(scores)[::-1][:top_n]

        for idx in top_indices: 
            if scores[idx] <= 0:
                break

            results[metric_name].append({
                "index": idx,
                "score": scores[idx],
                "matched_question": questions[idx],
                "answer": answers[idx]
            })
    return results
```

```python
import pandas as pd 

metric_names = ["Жаккар", "Косинусное сходство", "Корреляция Пирсона"]

evaluation_table = []
is_correct = {name: [] for name in metric_names}

for i, query in enumerate(test_questions):
    if i < len(correct_indices): 
        expected_index = correct_indices[i]
        correct_answer = answers[expected_index]
    else:
        expected_index = None
        correct_answer = "Ответ отсутствует в базе"

    results = find_answers(query, top_n=1)

    jaccard_result = results["Жаккар"][0]
    cosine_result = results["Косинусное сходство"][0]
    pearson_result = results["Корреляция Пирсона"][0]

    # Для вопросов без ответа в базе верным считаем отказ ("Ответ не найден")
    for name in metric_names:
        is_correct[name].append(results[name][0]["index"] == expected_index)

    evaluation_table.append({
        "Вопрос": query,
        "Правильный ответ": correct_answer,
        
        "Ответ системы (Жаккар)": jaccard_result["answer"],
        "Метрика Жаккара": round(jaccard_result["score"], 4),
        "Верно (Жаккар)": "да" if is_correct["Жаккар"][-1] else "нет",
        
        "Ответ системы (косинус)": cosine_result["answer"],
        "Косинусная близость": round(cosine_result["score"], 4),
        "Верно (косинус)": "да" if is_correct["Косинусное сходство"][-1] else "нет",
        
        "Ответ системы (корреляция)": pearson_result["answer"],
        "Корреляция Пирсона": round(pearson_result["score"], 4),
        "Верно (корреляция)": "да" if is_correct["Корреляция Пирсона"][-1] else "нет"
    })

evaluation_df = pd.DataFrame(evaluation_table)

evaluation_df
```

```python
n_in_base = len(correct_indices)
n_out_of_base = len(test_questions) - n_in_base

summary = []

for name in metric_names:
    flags = is_correct[name]
    in_base = sum(flags[:n_in_base])
    out_of_base = sum(flags[n_in_base:])

    summary.append({
        "Метрика": name,
        f"Верно, вопросы из базы (из {n_in_base})": in_base,
        f"Верно, вопросы вне базы (из {n_out_of_base})": out_of_base,
        f"Всего верно (из {len(test_questions)})": in_base + out_of_base,
        "Доля верных": round((in_base + out_of_base) / len(test_questions), 3)
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
# Разбор: что осталось от запроса после предобработки, какие слова
# есть в словаре базы и какой вопрос базы выбрала каждая метрика
vocabulary = set(feature_names)

details = []

for query in test_questions:
    tokens = preprocess_text(query)
    results = find_answers(query, top_n=1)

    row = {
        "Вопрос": query,
        "Токены": tokens,
        "Нет в словаре": [t for t in tokens if t not in vocabulary]
    }
    for name in metric_names:
        row[name] = results[name][0]["matched_question"] or NO_ANSWER
    details.append(row)

pd.DataFrame(details)
```

Все три метрики выбрали один и тот же вопрос базы во всех 15 случаях. Числа у них разные, порядок кандидатов одинаковый, поэтому по качеству ответов на этих данных метрики не различаются.

### Почему метрики совпали

Вопросы в базе короткие (2–5 лемм после обработки), и ни одно слово внутри вопроса не повторяется: максимальное значение в матрице мешка слов равно 1. Векторы получились бинарными, и частоты, на которых косинус мог бы выиграть, здесь не участвуют.

Если в запросе $q$ слов, в вопросе базы $d$ слов и $c$ из них общие, то

$$J = \frac{c}{q + d - c}, \qquad \cos = \frac{c}{\sqrt{q \cdot d}}.$$

Для фиксированного запроса обе величины растут с $c$ и падают с $d$, поэтому на коротких вопросах они упорядочивают кандидатов одинаково. Корреляция Пирсона отличается от косинуса только вычитанием среднего из каждого вектора. В векторе длины 59 с двумя-пятью единицами среднее близко к нулю, и корреляция почти совпадает с косинусом (например, 0.7071 и 0.7010, 0.5000 и 0.4869).

Разница между Жаккаром и косинусом видна в значениях. Жаккар считает все слова запроса, включая те, которых нет в словаре базы, а косинус их не видит: вектор запроса строится только по словарю. В строке 0 запрос «Сколько республик было в Советском Союзе на момент распада?» даёт 5 лемм, из них 3 вне словаря. Жаккар получает 1/7 ≈ 0.14, косинус 1/√6 ≈ 0.41. На выбор ответа это не повлияло: незнакомые слова одинаково увеличивают знаменатель для всех кандидатов. Но при выборе порога отсечения (пункт 3.3) Жаккар честнее показывает, что запрос понят плохо.

Косинусное расстояние из задания равно 1 − косинусное сходство. В ноутбуке считается сходство, чтобы у всех трёх метрик «больше» означало «ближе».

### Почему система ошибается

**Синонимы.** Словарь построен только по вопросам базы, а в них страна везде называется «СССР». В тестовых вопросах 1, 3 и 4 стоят «Советский Союз», «Германия», «Горбачёв», «реформа». Ни одного из этих слов нет в словаре, вектор запроса нулевой, и система отвечает «Ответ не найден», хотя ответ в базе есть. Мешок слов не знает, что «Советский Союз» и «СССР» одно и то же. Помогли бы расширение запроса синонимами (шаг 5 схемы IR-системы) или индексирование вместе с вопросами ещё и ответов: Горбачёв и нападение Германии там упомянуты.

**Совпадение по случайному слову.** Вопрос 2 «Кто возглавлял КПСС накануне распада Союза?» совпал с «Какое событие начало распад СССР?» по слову «распад». Нужный вопрос базы («Кто был последним секретарем партии в СССР?») не имеет с запросом общих слов.

**Слово «ссср».** Оно есть в 18 вопросах базы из 28, и в мешке слов весит столько же, сколько «коллективизация» или «спутник». Вопросы «Какой гимн был у СССР?» и «Сколько стоила буханка хлеба в СССР?» после обработки сводятся к «ссср» плюс незнакомые слова, и система уверенно отвечает про распад СССР (косинус 0.71). Конкретный вопрос выбран случайно: у нескольких кандидатов одинаковое значение, и `argsort` берёт последний. Вопрос про школьную систему совпал с вопросом про экономику по словам «устроить» и «ссср». Такие ошибки исправляют TF-IDF (понижает вес слов, которые встречаются почти везде) и порог минимального сходства.

### Жаккар на коротких текстах, косинус на длинных

Жаккар смотрит только на наличие слова: текст для него — множество. В коротком вопросе каждое слово встречается один раз, и множество теряет мало информации. Жаккар к тому же штрафует за каждое лишнее слово в объединении, что на коротких запросах удобно: два-три слова решают всё, и отличие в одном слове сразу заметно.

В длинном тексте слова повторяются, и частота показывает, о чём текст. Абзац, где «война» встречается шесть раз, посвящён войне, а абзац с одним упоминанием нет. Жаккар эту разницу выбрасывает. Объединение двух длинных текстов к тому же быстро разрастается, и коэффициент проседает даже у близких по смыслу текстов. Косинус использует частоты (а с TF-IDF ещё и редкость слова) и делит на длины векторов, поэтому длинный документ не выигрывает только за счёт объёма.

Наши результаты с этим согласуются: на коротких вопросах частоты ничего не добавили, и все метрики дали одинаковые ответы. Для базы из коротких вопросов достаточно Жаккара: он проще, точность та же, а значение лучше подходит для порога. Косинус имеет смысл брать, если сопоставлять запрос не с вопросами, а с длинными ответами, или после перехода на TF-IDF.
